```python
import pandas as pd
from deep_translator import GoogleTranslator
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import CountVectorizer

# CSV dosyasını oku
data = pd.read_csv("hindi.csv")

# Çeviriler için bir sözlük oluştur
translations = {}

# Her sütundaki benzersiz öğeleri çevir
for column in data.columns:
    unique = data[column].unique()
    for element in unique:
        try:
            translations[element] = GoogleTranslator(source='hi', target='en').translate(element)
        except Exception as e:
            print(f"'{element}' çevirisinde hata: {e}")
            translations[element] = "Çeviri Hatası"

# Çevirileri bir DataFrame'e dönüştür
translated_df = pd.DataFrame(list(translations.items()), columns=['Original', 'Translated'])

# Çevirilen verileri bir CSV dosyasına kaydet
translated_df.to_csv("translated_hindi.csv", index=False)

# Çevirilen verileri kontrol et
print(translated_df.head())
print(translated_df['Translated'].value_counts())

# Etiketleri oluşturma fonksiyonu
def label_data(text):
    if "good" in text:  # Olumlu anahtar kelime
        return 1
    elif "bad" in text:  # Olumsuz anahtar kelime
        return 2
    else:
        return 0  # Diğer durumlar

# Etiketleri oluştur
y = [label_data(text) for text in translated_df['Translated']]

# Etiketlerin sayısını kontrol et
print("Etiketlerin dağılımı:", pd.Series(y).value_counts())

# Eğer yalnızca bir etiket varsa, hata meydana gelmesin diye kontrol ekleyin
if len(set(y)) < 2:
    raise ValueError("Veri setinde en az iki farklı sınıf bulunmalıdır.")

# Modeli eğit
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(translated_df['Translated'])
model = LogisticRegression()
model.fit(X, y)

# Modeli kaydet
joblib.dump(model, 'translation_model.pkl')
joblib.dump(vectorizer, 'vectorizer.pkl')

print("Model ve vektörleştirici başarıyla kaydedildi.")
```
