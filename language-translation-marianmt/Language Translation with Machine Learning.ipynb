```python
import pandas as pd
from datasets import Dataset
from transformers import MarianMTModel, MarianTokenizer
from transformers import Trainer, TrainingArguments
```

```python
df = pd.read_csv('data.csv')
```

```python
df.head()
```

```python
import pandas as pd
from datasets import Dataset
from transformers import MarianMTModel, MarianTokenizer, Trainer, TrainingArguments

# Veri yükleme
data = {
    "english_sentence": df['english_sentence'],
    "hindi_sentence": df['hindi_sentence']
}
df = pd.DataFrame(data)

# Dataset oluşturma
dataset = Dataset.from_pandas(df)

# Model ve Tokenizer'ı yükleme
model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Tokenizasyon fonksiyonu
def preprocess_function(examples):
    inputs = examples['english_sentence']
    targets = examples['hindi_sentence']
    model_inputs = tokenizer(inputs, max_length=128, truncation=True)
    
    # Target tokenizasyonu
    with tokenizer.as_target_tokenizer():
        labels = tokenizer(targets, max_length=128, truncation=True)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# Veriyi işleme
tokenized_dataset = dataset.map(preprocess_function, batched=True)

# Eğitim argümanları
training_args = TrainingArguments(
    output_dir="./results",
    evaluation_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=4,
    num_train_epochs=3,
    logging_dir='./logs',  # Log dosyaları için bir dizin belirtebilirsiniz.
)

# Trainer oluşturma
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
)

# Modeli eğitme
trainer.train()

# Modeli kaydetme
model.save_pretrained("./hindi_translation_model")
tokenizer.save_pretrained("./hindi_translation_model")
```
